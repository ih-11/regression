```python
# ============================================
# Global plotting configuration
# ============================================

_PLOT_CFG = {
    "fig_w": 6.0,
    "fig_h": 6.0,
    "dpi": 300,
}


SPECIES_INFO = {
    "AT21": {
        "label": "AT",
        "short": "AT21",
        "color": "#664D0AFF",
        "marker": "o",
    },
    "NB21": {
        "label": "NB",
        "short": "NB21",
        "color": "#7e3131",
        "marker": "^",
    },
    "OS21": {
        "label": "OS",
        "short": "OS21",
        "color": "#13563f",
        "marker": "s",
    },
}

def set_plot_style(
    *,
    base_fontsize=10,
    title_fontsize=12,
    label_fontsize=10,
    tick_fontsize=9,
    legend_fontsize=10,
    dpi=300,
    axes_linewidth=1.2,
    spines_top=True,
    spines_right=True,
    tick_size_major=6,
    tick_dir="out",
    grid=False,
    fig_w=6.0,
    fig_h=6.0,
):
    sns.set_style("ticks")

    mpl.rcParams.update({
        "font.family": "DejaVu Sans",
        "font.size": base_fontsize,

        "axes.titlesize": title_fontsize,
        "axes.labelsize": label_fontsize,

        "xtick.labelsize": tick_fontsize,
        "ytick.labelsize": tick_fontsize,

        "legend.fontsize": legend_fontsize,

        "figure.dpi": dpi,
        "savefig.dpi": dpi,

        "axes.linewidth": axes_linewidth,
        "axes.spines.top": spines_top,
        "axes.spines.right": spines_right,
        "axes.grid": grid,
        "axes.axisbelow": True,

        "xtick.major.size": tick_size_major,
        "ytick.major.size": tick_size_major,
        "xtick.direction": tick_dir,
        "ytick.direction": tick_dir,

        "legend.frameon": False,

        "savefig.bbox": "tight",
        "savefig.transparent": False,
        "figure.autolayout": False,
    })

    _PLOT_CFG.update({
        "fig_w": fig_w,
        "fig_h": fig_h,
        "dpi": dpi,
    })


def make_fig(w=None, h=None, dpi=None):
    W = float(w) if w is not None else _PLOT_CFG["fig_w"]
    H = float(h) if h is not None else _PLOT_CFG["fig_h"]
    D = dpi if dpi is not None else _PLOT_CFG["dpi"]

    fig, ax = plt.subplots(
        figsize=(W, H),
        dpi=D,
    )

    return fig, ax


def _compact_formatter():
    def _fmt(x, _pos=None):
        axx = abs(x)

        if axx >= 1e9:
            s = f"{x / 1e9:.1f}B"
        elif axx >= 1e6:
            s = f"{x / 1e6:.1f}M"
        elif axx >= 1e3:
            s = f"{x / 1e3:.1f}k"
        else:
            s = f"{x:.2g}"

        return (
            s.replace(".0B", "B")
             .replace(".0M", "M")
             .replace(".0k", "k")
        )

    return FuncFormatter(_fmt)


def format_axis(
    ax,
    *,
    xlabel=None,
    ylabel=None,
    compact_ticks=(),
):
    if xlabel is not None:
        ax.set_xlabel(xlabel)

    if ylabel is not None:
        ax.set_ylabel(ylabel)

    fmt = _compact_formatter()

    if "x" in compact_ticks:
        ax.xaxis.set_major_formatter(fmt)

    if "y" in compact_ticks:
        ax.yaxis.set_major_formatter(fmt)

    return ax


# ============================================
# Joint scatter with KDE marginals
# ============================================

def safe_pearsonr(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if len(x) < 2:
        return np.nan, np.nan

    if np.std(x) == 0 or np.std(y) == 0:
        return np.nan, np.nan

    return stats.pearsonr(x, y)


def safe_spearmanr(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if len(x) < 2:
        return np.nan, np.nan

    if np.std(x) == 0 or np.std(y) == 0:
        return np.nan, np.nan

    return stats.spearmanr(x, y)

def _kde_1d(values, lo, hi, num=256):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    grid = np.linspace(lo, hi, num)

    if len(values) < 2:
        return grid, np.zeros_like(grid)

    try:
        kde = stats.gaussian_kde(values)
        dens = kde(grid)
        dens /= dens.max() if dens.max() > 0 else 1
        return grid, dens

    except Exception:
        return grid, np.zeros_like(grid)


def joint_scatter(
    x,
    y,
    *,
    color=None,
    point_size=18,
    alpha=0.65,
    show_identity=True,
    show_regression=True,
    annotate=True,
    annotate_spearman=True,
    xlabel=None,
    ylabel=None,
    title=None,
    figsize=None,
    w=None,
    h=None,
    dpi=None,
    annotate_fontsize=14,
):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    n = len(x)

    if figsize is not None:
        FW, FH = figsize
    else:
        FW = float(w) if w is not None else _PLOT_CFG["fig_w"]
        FH = float(h) if h is not None else _PLOT_CFG["fig_h"]

    fig = plt.figure(
        figsize=(FW, FH),
        dpi=(dpi or _PLOT_CFG["dpi"]),
    )

    gs = GridSpec(
        2,
        2,
        width_ratios=(4, 1),
        height_ratios=(1, 4),
        hspace=0.05,
        wspace=0.05,
    )

    ax_top = fig.add_subplot(gs[0, 0])
    ax_joint = fig.add_subplot(gs[1, 0], sharex=ax_top)
    ax_right = fig.add_subplot(gs[1, 1], sharey=ax_joint)

    ax_joint.scatter(
        x,
        y,
        s=point_size,
        alpha=alpha,
        edgecolor="none",
        color=color,
    )

    lo = float(np.nanmin([x.min(), y.min()]))
    hi = float(np.nanmax([x.max(), y.max()]))

    pad = 0.05 * (hi - lo if hi > lo else 1.0)

    lo -= pad
    hi += pad

    ax_joint.set_xlim(lo, hi)
    ax_joint.set_ylim(lo, hi)

    if show_identity:
        ax_joint.plot(
            [lo, hi],
            [lo, hi],
            ls="--",
            lw=1.2,
            color="0.65",
            zorder=1,
        )

    if show_regression and n >= 2 and np.std(x) > 0 and np.std(y) > 0:
        slope, intercept = np.polyfit(x, y, 1)

        ax_joint.plot(
            [lo, hi],
            slope * np.array([lo, hi]) + intercept,
            color="black",
            lw=1.5,
            zorder=2,
        )

    format_axis(
        ax_joint,
        xlabel=xlabel,
        ylabel=ylabel,
        compact_ticks=(),
    )

    if title:
        ax_joint.set_title(title)

    if annotate and n >= 2 and np.std(x) > 0 and np.std(y) > 0:
        rp, _ = safe_pearsonr(x, y)
        rs, _ = safe_spearmanr(x, y)

        txt = rf"$r_p = {rp:.2f}$"

        if annotate_spearman:
            txt += "\n" + rf"$r_s = {rs:.2f}$"

        txt += f"\n$n = {n}$"

        ax_joint.text(
            0.04,
            0.96,
            txt,
            transform=ax_joint.transAxes,
            ha="left",
            va="top",
            fontsize=annotate_fontsize,
        )

    gx, dx = _kde_1d(x, lo, hi)
    gy, dy = _kde_1d(y, lo, hi)

    ax_top.plot(gx, dx, lw=2, color=color)
    ax_top.axis("off")

    ax_right.plot(dy, gy, lw=2, color=color)
    ax_right.axis("off")

    plt.tight_layout()

    return fig, (ax_joint, ax_top, ax_right)


set_plot_style()

# ============================================
# Plot helper functions
# ============================================

def species_palette():
    return [
        SPECIES_INFO["AT21"]["color"],
        SPECIES_INFO["NB21"]["color"],
        SPECIES_INFO["OS21"]["color"],
    ]


def outside_legend(
    ax,
    *,
    title="Species",
    n_items=3,
    x=1.02,
    y=1.00,
):
    handles, labels = ax.get_legend_handles_labels()

    ax.legend(
        handles[:n_items],
        labels[:n_items],
        title=title,
        frameon=False,
        loc="upper left",
        bbox_to_anchor=(x, y),
        borderaxespad=0,
    )

    return ax
```

```python
# ============================================
# 01. Load locked selected-feature dataset
# ============================================

RESULT_DIR = Path("/mnt/d/Ibnu/Programming/data/regression/results/lgbm")
SPECIES_KEYS = ["AT21", "NB21", "OS21"]
SPECIES_ORDER = ["AT", "NB", "OS"]

# Locked FI + SSS selection table
candidate_all_df = pd.read_csv(
    RESULT_DIR / "lgbm.species_specific_candidate_table.importance_p90.sss_p90.tsv.gz",
    sep="\t",
)

candidate_selected_df = (
    candidate_all_df
    .loc[candidate_all_df["Selected"]]
    .copy()
    .reset_index(drop=True)
)

candidate_features = sorted(candidate_selected_df["Feature"].unique())

feature_metadata_df = (
    candidate_selected_df[["Feature", "Region", "Feature_type"]]
    .drop_duplicates()
    .sort_values("Feature")
    .reset_index(drop=True)
)

# Raw biological data for the 61-feature union
raw_rows = []

for species_key in SPECIES_KEYS:
    species = SPECIES_INFO[species_key]["label"]

    with gzip.open(
        RESULT_DIR / f"{species_key}.lgbm.data_bundle.pkl.gz",
        "rb",
    ) as f:
        bundle = pickle.load(f)

    assert bundle["species"] == species_key
    assert set(candidate_features).issubset(bundle["feature_names"])

    metadata = pd.concat(
        [bundle["train_metadata"], bundle["test_metadata"]]
    )

    x_raw = pd.concat(
        [bundle["x_train_raw"], bundle["x_test_raw"]]
    )[candidate_features]

    assert metadata.index.equals(x_raw.index)

    temp = pd.concat([metadata, x_raw], axis=1)
    temp.insert(0, "Species", species)

    raw_rows.append(temp)

candidate_master_df = pd.concat(
    raw_rows,
    ignore_index=True,
)

# Sanity checks
selected_count_df = (
    candidate_selected_df
    .groupby("Species")
    .size()
    .reindex(SPECIES_ORDER)
    .rename("N_selected")
    .reset_index()
)

assert selected_count_df.set_index("Species")["N_selected"].to_dict() == {
    "AT": 19,
    "NB": 28,
    "OS": 23,
}

assert len(candidate_features) == 61
assert feature_metadata_df["Feature"].is_unique
assert set(feature_metadata_df["Feature"]) == set(candidate_features)
assert candidate_master_df[candidate_features].isna().sum().sum() == 0

print("Locked selected-feature dataset")
print("--------------------------------")
print("Selected rows   :", len(candidate_selected_df))
print("Unique features :", len(candidate_features))
print("Raw data shape  :", candidate_master_df.shape)

display(selected_count_df)
display(feature_metadata_df.head())
```

```python
# ============================================
# 02. Predictive selection membership
# SSF / PSF / MSF
# ============================================

membership_df = (
    candidate_selected_df.assign(Selected=1)
    .pivot_table(index="Feature", columns="Species", values="Selected", aggfunc="max", fill_value=0)
    .reindex(columns=SPECIES_ORDER, fill_value=0).reset_index()
    .merge(feature_metadata_df, on="Feature", how="left", validate="one_to_one")
)

membership_df["N_species"] = membership_df[SPECIES_ORDER].sum(axis=1).astype(int)
membership_df["Selection_class"] = membership_df["N_species"].map({1: "SSF", 2: "PSF", 3: "MSF"})
membership_df["Membership"] = membership_df.apply(
    lambda r: "-".join(sp for sp in SPECIES_ORDER if r[sp] == 1), axis=1
)

membership_df = (
    membership_df[
        ["Feature", "Region", "Feature_type", "AT", "NB", "OS",
         "N_species", "Selection_class", "Membership"]
    ]
    .sort_values(["N_species", "Membership", "Feature"], ascending=[False, True, True])
    .reset_index(drop=True)
)

class_summary_df = (
    membership_df["Selection_class"].value_counts()
    .reindex(["SSF", "PSF", "MSF"], fill_value=0)
    .rename("N_features").rename_axis("Selection_class").reset_index()
)

membership_summary_df = (
    membership_df.groupby(["Selection_class", "Membership"], as_index=False)
    .agg(N_features=("Feature", "size"))
)

assert len(membership_df) == 61
assert membership_df["Feature"].is_unique
assert membership_df["N_species"].between(1, 3).all()

print("Predictive selection membership")
display(class_summary_df)
display(membership_summary_df)
display(membership_df)
```

```python
# ============================================
# 03. Predictive vocabulary composition
# Region + feature type
# ============================================

region_order = ["5'UTR", "CDS", "3'UTR", "mRNA"]
type_order = ["Length", "RNA_structure_MFE", "Amino_acid_composition", "Nucleotide_kmer_freq"]
class_order = ["SSF", "PSF", "MSF"]

region_summary_df = (
    membership_df.groupby(["Selection_class", "Region"], as_index=False)
    .agg(N_features=("Feature", "size"))
)
region_summary_df["Percent"] = (
    100 * region_summary_df["N_features"]
    / region_summary_df.groupby("Selection_class")["N_features"].transform("sum")
)

type_summary_df = (
    membership_df.groupby(["Selection_class", "Feature_type"], as_index=False)
    .agg(N_features=("Feature", "size"))
)
type_summary_df["Percent"] = (
    100 * type_summary_df["N_features"]
    / type_summary_df.groupby("Selection_class")["N_features"].transform("sum")
)

overall_summary_df = pd.DataFrame({
    "N_features": [len(membership_df)],
    "N_SSF": [(membership_df["Selection_class"] == "SSF").sum()],
    "N_PSF": [(membership_df["Selection_class"] == "PSF").sum()],
    "N_MSF": [(membership_df["Selection_class"] == "MSF").sum()],
})

print("Predictive vocabulary composition")
display(overall_summary_df)
display(region_summary_df.sort_values(["Selection_class", "Region"]))
display(type_summary_df.sort_values(["Selection_class", "Feature_type"]))
```

```python
# ============================================
# 06. Feature–PR association across species
# Spearman rho + BH-FDR
# ============================================

from statsmodels.stats.multitest import multipletests

rows = []

for species in SPECIES_ORDER:
    d = candidate_master_df.query("Species == @species")

    for feature in candidate_features:
        rho, p = safe_spearmanr(d[feature], d["observed_raw"])
        rows.append({"Species": species, "Feature": feature, "PR_rho": rho, "P_value": p})

feature_pr_df = pd.DataFrame(rows)

feature_pr_df["P_adj"] = (
    feature_pr_df.groupby("Species")["P_value"]
    .transform(lambda x: multipletests(x, method="fdr_bh")[1])
)

feature_pr_df["Significant"] = feature_pr_df["P_adj"] < 0.05

feature_pr_df = (
    feature_pr_df
    .merge(
        membership_df[["Feature", "Region", "Feature_type", "Selection_class", "Membership"]],
        on="Feature", how="left", validate="many_to_one"
    )
    .sort_values(["Species", "PR_rho"], ascending=[True, False])
    .reset_index(drop=True)
)

pr_summary_df = (
    feature_pr_df.groupby("Species", as_index=False)
    .agg(
        N_features=("Feature", "size"),
        N_significant=("Significant", "sum"),
        Median_abs_rho=("PR_rho", lambda x: x.abs().median()),
        Max_abs_rho=("PR_rho", lambda x: x.abs().max()),
    )
)

print("Feature–PR association")
display(pr_summary_df)
display(feature_pr_df)
```

```python
# ============================================
# 07. Cross-species PR effect matrix
# 61 features × 3 species
# ============================================

pr_effect_df = (
    feature_pr_df
    .pivot(index="Feature", columns="Species", values="PR_rho")
    .reindex(columns=SPECIES_ORDER)
    .reset_index()
    .merge(
        membership_df[["Feature", "Region", "Feature_type", "Selection_class", "Membership"]],
        on="Feature", how="left", validate="one_to_one"
    )
)

pr_q_df = (
    feature_pr_df
    .pivot(index="Feature", columns="Species", values="P_adj")
    .reindex(columns=SPECIES_ORDER)
    .reset_index()
    .rename(columns={s: f"{s}_q" for s in SPECIES_ORDER})
)

pr_effect_df = (
    pr_effect_df
    .merge(pr_q_df, on="Feature", validate="one_to_one")
    .rename(columns={s: f"{s}_rho" for s in SPECIES_ORDER})
)

pr_effect_df["Max_abs_rho"] = pr_effect_df[
    [f"{s}_rho" for s in SPECIES_ORDER]
].abs().max(axis=1)

pr_effect_df = (
    pr_effect_df
    .sort_values("Max_abs_rho", ascending=False)
    .reset_index(drop=True)
)

assert len(pr_effect_df) == 61
assert pr_effect_df["Feature"].is_unique

display(
    pr_effect_df[
        ["Feature", "Region", "Feature_type", "Selection_class", "Membership",
         "AT_rho", "NB_rho", "OS_rho", "AT_q", "NB_q", "OS_q", "Max_abs_rho"]
    ]
)
```

```python
# ============================================
# 08. Global cross-species PR-effect conservation
# ============================================

from itertools import combinations

pairs = list(combinations(SPECIES_ORDER, 2))
point_color = "#8F6F6F"
rows = []

for s1, s2 in pairs:
    rho, p = safe_spearmanr(pr_effect_df[f"{s1}_rho"], pr_effect_df[f"{s2}_rho"])
    rows.append({"Species_pair": f"{s1}-{s2}", "N_features": len(pr_effect_df),
                 "Spearman_rho": rho, "P_value": p})

pr_effect_conservation_df = pd.DataFrame(rows)
display(pr_effect_conservation_df)

vals = pr_effect_df[[f"{s}_rho" for s in SPECIES_ORDER]].to_numpy(dtype=float)
lim = 1.08 * np.nanmax(np.abs(vals))

fig, axes = plt.subplots(
    1, 3, figsize=(7.5, 2.6), dpi=_PLOT_CFG["dpi"],
    sharex=True, sharey=True,
)

for ax, (s1, s2) in zip(axes, pairs):
    x, y = pr_effect_df[f"{s1}_rho"], pr_effect_df[f"{s2}_rho"]
    rho, p = safe_spearmanr(x, y)

    ax.scatter(x, y, s=20, color=point_color, alpha=0.75)
    ax.axhline(0, color="0.82", lw=0.8)
    ax.axvline(0, color="0.82", lw=0.8)

    p_text = f"{p:.2e}" if p < 0.001 else f"{p:.3f}"
    ax.text(
        0.05, 0.94,
        rf"Spearman $\rho$ = {rho:.2f}" + "\n" + rf"$p$ = {p_text}",
        transform=ax.transAxes, ha="left", va="top", fontsize=7.5,
    )

    ax.set_xlim(-lim, lim)
    ax.set_ylim(-lim, lim)
    ax.set_title(f"{s1} vs {s2}")
    ax.set_xlabel(f"{s1} PR effect")
    ax.set_ylabel(f"{s2} PR effect")

plt.tight_layout()
plt.show()
```

```python
# ============================================
# 09. Feature-level PR-effect divergence
# ============================================

rho_cols = [f"{s}_rho" for s in SPECIES_ORDER]

pr_divergence_df = pr_effect_df.copy()
pr_divergence_df["Effect_range"] = (
    pr_divergence_df[rho_cols].max(axis=1) - pr_divergence_df[rho_cols].min(axis=1)
)
pr_divergence_df["Effect_SD"] = pr_divergence_df[rho_cols].std(axis=1)
pr_divergence_df["Strongest_species"] = (
    pr_divergence_df[rho_cols].abs().idxmax(axis=1).str.replace("_rho", "", regex=False)
)
pr_divergence_df["Direction"] = pr_divergence_df[rho_cols].apply(
    lambda r: "Same" if len(set(np.sign(r))) == 1 else "Mixed", axis=1
)

pr_divergence_df = (
    pr_divergence_df
    .sort_values("Effect_range", ascending=False)
    .reset_index(drop=True)
)

display(
    pr_divergence_df[
        ["Feature", "Region", "Feature_type", "Selection_class", "Membership",
         "AT_rho", "NB_rho", "OS_rho", "Effect_range", "Effect_SD",
         "Strongest_species", "Direction"]
    ]
)
```

```python
# ============================================
# 10. Bootstrap PR effects and differences
# ============================================

N_BOOT = 2000
rng = np.random.default_rng(42)
effect_rows, diff_rows = [], []
rho_lookup = pr_effect_df.set_index("Feature")

for feature in candidate_features:
    boot = {}

    for sp in SPECIES_ORDER:
        d = candidate_master_df.loc[
            candidate_master_df["Species"].eq(sp), [feature, "observed_raw"]
        ].dropna().to_numpy()
        n = len(d)
        boot[sp] = np.array([
            safe_spearmanr(*d[rng.integers(0, n, n)].T)[0] for _ in range(N_BOOT)
        ])

        effect_rows.append({
            "Feature": feature, "Species": sp,
            "Rho": rho_lookup.loc[feature, f"{sp}_rho"],
            "CI_low": np.nanquantile(boot[sp], 0.025),
            "CI_high": np.nanquantile(boot[sp], 0.975),
        })

    for s1, s2 in combinations(SPECIES_ORDER, 2):
        delta = boot[s1] - boot[s2]
        diff_rows.append({
            "Feature": feature, "Species_pair": f"{s1}-{s2}",
            "Delta_rho": rho_lookup.loc[feature, f"{s1}_rho"] - rho_lookup.loc[feature, f"{s2}_rho"],
            "CI_low": np.nanquantile(delta, 0.025),
            "CI_high": np.nanquantile(delta, 0.975),
        })

pr_effect_ci_df = pd.DataFrame(effect_rows)
pr_difference_df = pd.DataFrame(diff_rows)

for df in [pr_effect_ci_df, pr_difference_df]:
    df["CI_excludes_zero"] = (df["CI_low"] > 0) | (df["CI_high"] < 0)

display(pr_effect_ci_df.head())
display(pr_difference_df.head())
```

```python
# ============================================
# 11. Integrate PR-effect evidence
# ============================================

effect_wide = (
    pr_effect_ci_df
    .pivot(index="Feature", columns="Species",
           values=["Rho", "CI_low", "CI_high", "CI_excludes_zero"])
)
effect_wide.columns = [f"{sp}_{metric}" for metric, sp in effect_wide.columns]
effect_wide = effect_wide.reset_index()

diff_wide = (
    pr_difference_df
    .pivot(index="Feature", columns="Species_pair",
           values=["Delta_rho", "CI_low", "CI_high", "CI_excludes_zero"])
)
diff_wide.columns = [f"{pair}_{metric}" for metric, pair in diff_wide.columns]
diff_wide = diff_wide.reset_index()

pr_evidence_df = (
    membership_df[["Feature", "Region", "Feature_type", "Selection_class", "Membership"]]
    .merge(effect_wide, on="Feature", validate="one_to_one")
    .merge(diff_wide, on="Feature", validate="one_to_one")
)

effect_support = [f"{s}_CI_excludes_zero" for s in SPECIES_ORDER]
diff_support = [f"{p}_CI_excludes_zero" for p in ["AT-NB", "AT-OS", "NB-OS"]]

pr_evidence_df["N_supported_species"] = pr_evidence_df[effect_support].sum(axis=1).astype(int)
pr_evidence_df["N_different_pairs"] = pr_evidence_df[diff_support].sum(axis=1).astype(int)
pr_evidence_df["Effect_range"] = (
    pr_evidence_df[[f"{s}_Rho" for s in SPECIES_ORDER]].max(axis=1)
    - pr_evidence_df[[f"{s}_Rho" for s in SPECIES_ORDER]].min(axis=1)
)

pr_evidence_df = (
    pr_evidence_df
    .sort_values(["N_different_pairs", "Effect_range"], ascending=False)
    .reset_index(drop=True)
)

display(
    pr_evidence_df[
        ["Feature", "Selection_class", "Membership",
         "AT_Rho", "NB_Rho", "OS_Rho",
         "N_supported_species", "N_different_pairs", "Effect_range"]
    ]
)
```

```python
# ============================================
# 13. Cross-species PR-effect direction
# ============================================

rho_cols = [f"{s}_Rho" for s in SPECIES_ORDER]

def sign_pattern(r):
    x = r[rho_cols].to_numpy(float)
    return "".join("+" if v > 0 else "-" if v < 0 else "0" for v in x)

pr_evidence_df["Sign_pattern"] = pr_evidence_df.apply(sign_pattern, axis=1)
pr_evidence_df["Same_direction"] = (
    pr_evidence_df[rho_cols].gt(0).all(axis=1)
    | pr_evidence_df[rho_cols].lt(0).all(axis=1)
)

direction_summary_df = (
    pr_evidence_df.groupby(["Same_direction", "Sign_pattern"])
    .size().rename("N_features").reset_index()
    .sort_values("N_features", ascending=False)
)

direction_class_df = (
    pr_evidence_df["Same_direction"]
    .map({True: "Same direction", False: "Mixed direction"})
    .value_counts()
    .reindex(["Same direction", "Mixed direction"], fill_value=0)
    .rename("N_features").rename_axis("Direction_class").reset_index()
)
direction_class_df["Percent"] = 100 * direction_class_df["N_features"] / len(pr_evidence_df)

display(direction_class_df)
display(direction_summary_df)

fig, axes = plt.subplots(1, 2, figsize=(7, 2.8), dpi=_PLOT_CFG["dpi"])

axes[0].bar(direction_class_df["Direction_class"], direction_class_df["N_features"])
axes[0].set(ylabel="Number of features", xlabel="")
axes[0].tick_params(axis="x", rotation=20)

p = direction_summary_df.copy()
axes[1].bar(p["Sign_pattern"], p["N_features"])
axes[1].set(xlabel="AT / NB / OS sign pattern", ylabel="Number of features")
axes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()

display(
    pr_evidence_df[
        ["Feature", "Selection_class", "Membership",
         "AT_Rho", "NB_Rho", "OS_Rho",
         "Sign_pattern", "Same_direction",
         "N_supported_species", "N_different_pairs", "Effect_range"]
    ]
    .sort_values(["Same_direction", "Effect_range"], ascending=[True, False])
    .reset_index(drop=True)
)
```

```python
# ============================================
# 14. PR-effect support patterns
# ============================================

def support_pattern(r):
    return "-".join(s for s in SPECIES_ORDER if r[f"{s}_CI_excludes_zero"]) or "None"

# ============================================
# 14. PR-effect support patterns
# ============================================

pairs = ["AT-NB", "AT-OS", "NB-OS"]

pr_evidence_df["Supported_species"] = pr_evidence_df.apply(
    lambda r: "-".join(s for s in SPECIES_ORDER if r[f"{s}_CI_excludes_zero"]) or "None", axis=1
)
pr_evidence_df["Different_pairs"] = pr_evidence_df.apply(
    lambda r: " + ".join(p for p in pairs if r[f"{p}_CI_excludes_zero"]) or "None", axis=1
)

support_order = ["AT-NB-OS", "AT-NB", "AT-OS", "NB-OS", "AT", "NB", "OS", "None"]
diff_order = [
    "AT-NB + AT-OS + NB-OS", "AT-NB + AT-OS", "AT-NB + NB-OS",
    "AT-OS + NB-OS", "AT-NB", "AT-OS", "NB-OS", "None"
]

def summarize_pattern(col, order):
    out = (
        pr_evidence_df[col].value_counts()
        .reindex(order, fill_value=0)
        .rename("N_features").rename_axis(col).reset_index()
    )
    out["Percent"] = 100 * out["N_features"] / len(pr_evidence_df)
    return out.loc[out["N_features"] > 0].reset_index(drop=True)

support_summary_df = summarize_pattern("Supported_species", support_order)
difference_summary_df = summarize_pattern("Different_pairs", diff_order)

display(support_summary_df)
display(difference_summary_df)

fig, axes = plt.subplots(1, 2, figsize=(7, 3.2), dpi=_PLOT_CFG["dpi"])

for ax, df, col, title in [
    (axes[0], support_summary_df, "Supported_species", "Supported PR effects"),
    (axes[1], difference_summary_df, "Different_pairs", "Cross-species differences"),
]:
    y = np.arange(len(df))
    ax.barh(y, df["N_features"])
    ax.set_yticks(y, df[col])
    ax.invert_yaxis()
    ax.set(title=title, xlabel="Number of features", ylabel="")
    for i, (n, pct) in enumerate(zip(df["N_features"], df["Percent"])):
        ax.text(n + 0.3, i, f"{n} ({pct:.1f}%)", va="center", fontsize=7)

plt.tight_layout()
plt.show()
```

```python
# ============================================
# 15. Joint PR-effect evidence
# ============================================

def evidence_pattern(r):
    n_sup, n_diff = r["N_supported_species"], r["N_different_pairs"]

    if n_sup == 3 and n_diff == 0 and r["Same_direction"]:
        return "Shared stable"
    if n_sup == 3 and n_diff > 0 and r["Same_direction"]:
        return "Shared divergent"
    if n_sup >= 2 and not r["Same_direction"]:
        return "Direction divergent"
    if n_sup == 2:
        return "Partial"
    if n_sup == 1:
        return "Restricted"
    return "Unsupported"

pr_evidence_df["Evidence_class"] = pr_evidence_df.apply(evidence_pattern, axis=1)

evidence_order = [
    "Shared stable", "Shared divergent", "Direction divergent",
    "Partial", "Restricted", "Unsupported",
]

evidence_summary_df = (
    pr_evidence_df["Evidence_class"].value_counts()
    .reindex(evidence_order, fill_value=0)
    .rename("N_features").rename_axis("Evidence_class").reset_index()
)
evidence_summary_df["Percent"] = 100 * evidence_summary_df["N_features"] / len(pr_evidence_df)

display(evidence_summary_df)

fig, ax = plt.subplots(figsize=(4.5, 3), dpi=_PLOT_CFG["dpi"])
ax.barh(evidence_summary_df["Evidence_class"], evidence_summary_df["N_features"])
ax.invert_yaxis()
ax.set(xlabel="Number of features", ylabel="", title="Joint PR-effect evidence")

for i, (n, p) in enumerate(zip(evidence_summary_df["N_features"], evidence_summary_df["Percent"])):
    ax.text(n + 0.2, i, f"{n} ({p:.1f}%)", va="center", fontsize=7)

plt.tight_layout()
plt.show()
```

```python
# ============================================
# 16. Joint evidence combinations
# ============================================

joint_summary_df = (
    pr_evidence_df
    .groupby(["N_supported_species", "N_different_pairs", "Same_direction"])
    .size().rename("N_features").reset_index()
)

joint_summary_df["Percent"] = 100 * joint_summary_df["N_features"] / len(pr_evidence_df)
joint_summary_df = joint_summary_df.sort_values(
    ["N_supported_species", "N_different_pairs", "Same_direction"],
    ascending=[False, True, False]
).reset_index(drop=True)

display(joint_summary_df)

joint_feature_df = (
    pr_evidence_df[
        ["Feature", "Region", "Feature_type", "Selection_class", "Membership",
         "AT_Rho", "NB_Rho", "OS_Rho", "Sign_pattern",
         "N_supported_species", "N_different_pairs", "Same_direction",
         "Effect_range"]
    ]
    .sort_values(["N_supported_species", "N_different_pairs", "Effect_range"],
                 ascending=[False, True, False])
    .reset_index(drop=True)
)

display(joint_feature_df)
```

```python
# ============================================
# 17. Supported PR-effect direction
# ============================================

def supported_sign(r, sp):
    if not r[f"{sp}_CI_excludes_zero"]:
        return "0"
    return "+" if r[f"{sp}_Rho"] > 0 else "-"

for sp in SPECIES_ORDER:
    pr_evidence_df[f"{sp}_Sign"] = pr_evidence_df.apply(
        lambda r: supported_sign(r, sp), axis=1
    )

pr_evidence_df["Supported_sign_pattern"] = pr_evidence_df[
    [f"{s}_Sign" for s in SPECIES_ORDER]
].agg("".join, axis=1)

def supported_direction(r):
    signs = [r[f"{s}_Sign"] for s in SPECIES_ORDER if r[f"{s}_Sign"] != "0"]
    if len(signs) < 2:
        return "Not comparable"
    return "Same" if len(set(signs)) == 1 else "Opposite"

pr_evidence_df["Supported_direction"] = pr_evidence_df.apply(
    supported_direction, axis=1
)

supported_direction_summary_df = (
    pr_evidence_df.groupby(["Supported_sign_pattern", "Supported_direction"])
    .size().rename("N_features").reset_index()
    .sort_values("N_features", ascending=False).reset_index(drop=True)
)

display(supported_direction_summary_df)
```

```python
# ============================================
# 18. Final PR regulatory classes
# ============================================

def regulatory_class(r):
    if r["N_supported_species"] == 3 and r["Supported_direction"] == "Same" and r["N_different_pairs"] == 0:
        return "Conserved"

    if r["Supported_direction"] == "Opposite":
        return "Opposite direction"

    if r["N_supported_species"] == 1:
        sp = next(s for s in SPECIES_ORDER if r[f"{s}_CI_excludes_zero"])
        others = [s for s in SPECIES_ORDER if s != sp]
        if all(r[f"{'-'.join(sorted([sp, x], key=SPECIES_ORDER.index))}_CI_excludes_zero"] for x in others):
            return "Species-restricted"

    if r["N_supported_species"] >= 2 and r["Supported_direction"] == "Same":
        return "Partially conserved"

    return "No clear pattern"

pr_evidence_df["PR_class"] = pr_evidence_df.apply(regulatory_class, axis=1)

class_order = [
    "Conserved", "Partially conserved", "Opposite direction",
    "Species-restricted", "No clear pattern",
]

pr_class_summary_df = (
    pr_evidence_df["PR_class"].value_counts()
    .reindex(class_order, fill_value=0)
    .rename("N_features").rename_axis("PR_class").reset_index()
)
pr_class_summary_df["Percent"] = 100 * pr_class_summary_df["N_features"] / len(pr_evidence_df)

display(pr_class_summary_df)
```

```python
# ============================================
# 19. Inspect final PR regulatory classes
# ============================================

cols = [
    "Feature", "Region", "Feature_type", "Selection_class", "Membership",
    "AT_Rho", "NB_Rho", "OS_Rho", "Supported_sign_pattern",
    "N_supported_species", "N_different_pairs", "Effect_range",
]

for cls in ["Conserved", "Species-restricted", "Opposite direction",
            "Partially conserved", "No clear pattern"]:
    x = (
        pr_evidence_df.loc[pr_evidence_df["PR_class"].eq(cls), cols]
        .sort_values("Effect_range", ascending=False).reset_index(drop=True)
    )
    print(f"\n{cls} (n={len(x)})")
    display(x)
```

```python
# ============================================
# 20. Refine PR regulatory classes
# ============================================

def regulatory_class(r):
    n = r["N_supported_species"]

    if n == 3 and r["Supported_direction"] == "Same":
        return "Conserved" if r["N_different_pairs"] == 0 else "Quantitatively divergent"

    if n >= 2 and r["Supported_direction"] == "Opposite":
        return "Opposite direction"

    if n == 2 and r["Supported_direction"] == "Same":
        return "Partially shared"

    if n == 1:
        sp = next(s for s in SPECIES_ORDER if r[f"{s}_CI_excludes_zero"])
        related = [p for p in ["AT-NB", "AT-OS", "NB-OS"] if sp in p.split("-")]
        if all(r[f"{p}_CI_excludes_zero"] for p in related):
            return "Species-restricted"

    return "No clear pattern"

pr_evidence_df["PR_class"] = pr_evidence_df.apply(regulatory_class, axis=1)

class_order = [
    "Conserved", "Quantitatively divergent", "Partially shared",
    "Opposite direction", "Species-restricted", "No clear pattern",
]

pr_class_summary_df = (
    pr_evidence_df["PR_class"].value_counts()
    .reindex(class_order, fill_value=0)
    .rename("N_features").rename_axis("PR_class").reset_index()
)
pr_class_summary_df["Percent"] = 100 * pr_class_summary_df["N_features"] / len(pr_evidence_df)

display(pr_class_summary_df)
```

```python
# ============================================
# 21. Predictive selection × PR response
# ============================================

selection_pr_df = pd.crosstab(
    pr_evidence_df["Selection_class"],
    pr_evidence_df["PR_class"]
).reindex(
    index=["MSF", "PSF", "SSF"],
    columns=class_order,
    fill_value=0,
)

selection_pr_pct = selection_pr_df.div(selection_pr_df.sum(axis=1), axis=0) * 100

display(selection_pr_df)
display(selection_pr_pct.round(1))

fig, axes = plt.subplots(1, 2, figsize=(8, 3), dpi=_PLOT_CFG["dpi"])

selection_pr_df.plot.bar(stacked=True, ax=axes[0], legend=False)
selection_pr_pct.plot.bar(stacked=True, ax=axes[1])

axes[0].set(xlabel="Selection class", ylabel="Number of features", title="Counts")
axes[1].set(xlabel="Selection class", ylabel="Percent", title="Within-class composition")
axes[1].legend(title="PR response", bbox_to_anchor=(1.02, 1), loc="upper left", frameon=False)

for ax in axes:
    ax.tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.show()
```

```python
# ============================================
# 22. SSF membership × PR response
# ============================================

ssf_df = pr_evidence_df.query("Selection_class == 'SSF'").copy()

ssf_pr_df = pd.crosstab(
    ssf_df["Membership"], ssf_df["PR_class"]
).reindex(index=SPECIES_ORDER, columns=class_order, fill_value=0)

ssf_pr_pct = ssf_pr_df.div(ssf_pr_df.sum(axis=1), axis=0) * 100

display(ssf_pr_df)
display(ssf_pr_pct.round(1))

fig, axes = plt.subplots(1, 2, figsize=(8, 3), dpi=_PLOT_CFG["dpi"])

ssf_pr_df.plot.bar(stacked=True, ax=axes[0], legend=False)
ssf_pr_pct.plot.bar(stacked=True, ax=axes[1])

axes[0].set(title="Counts", xlabel="SSF membership", ylabel="Number of features")
axes[1].set(title="Within-species composition", xlabel="SSF membership", ylabel="Percent")
axes[1].legend(title="PR response", bbox_to_anchor=(1.02, 1), loc="upper left", frameon=False)

for ax in axes:
    ax.tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.show()
```

```python
# ============================================
# 23. SSF selection–effect concordance
# ============================================

ssf_df = pr_evidence_df.query("Selection_class == 'SSF'").copy()
rho_cols = [f"{s}_Rho" for s in SPECIES_ORDER]

ssf_df["Strongest_species"] = (
    ssf_df[rho_cols].abs().idxmax(axis=1).str.replace("_Rho", "", regex=False)
)
ssf_df["Selection_concordant"] = ssf_df["Membership"].eq(ssf_df["Strongest_species"])

concordance_df = (
    ssf_df.groupby("Membership")["Selection_concordant"]
    .agg(N="size", N_concordant="sum").reset_index()
)
concordance_df["Percent_concordant"] = 100 * concordance_df["N_concordant"] / concordance_df["N"]

display(concordance_df)
display(
    ssf_df[["Feature", "Membership", "Strongest_species",
            "AT_Rho", "NB_Rho", "OS_Rho", "PR_class", "Selection_concordant"]]
    .sort_values(["Membership", "Selection_concordant"])
)
```

```python
# ============================================
# 24. Test SSF selection–effect concordance
# ============================================

from scipy.stats import binomtest

rows = []

for sp in ["All", *SPECIES_ORDER]:
    d = ssf_df if sp == "All" else ssf_df.query("Membership == @sp")
    k, n = d["Selection_concordant"].sum(), len(d)
    test = binomtest(k, n, p=1/3, alternative="greater")

    rows.append({
        "Membership": sp, "N": n, "N_concordant": k,
        "Percent": 100 * k / n, "P_value": test.pvalue,
    })

concordance_test_df = pd.DataFrame(rows)
concordance_test_df["FDR"] = multipletests(
    concordance_test_df["P_value"], method="fdr_bh"
)[1]

display(concordance_test_df)
```

```python
# ============================================
# 25. SSF concordance by PR-response class
# ============================================

concordance_class_df = (
    ssf_df
    .groupby(["Membership", "PR_class", "Selection_concordant"])
    .size()
    .unstack(fill_value=0)
    .rename(columns={False: "Discordant", True: "Concordant"})
)

for col in ["Discordant", "Concordant"]:
    if col not in concordance_class_df.columns:
        concordance_class_df[col] = 0

concordance_class_df["N"] = (
    concordance_class_df["Concordant"]
    + concordance_class_df["Discordant"]
)

concordance_class_df["Percent_concordant"] = (
    100
    * concordance_class_df["Concordant"]
    / concordance_class_df["N"]
)

concordance_class_df = (
    concordance_class_df
    .reset_index()
    .sort_values(
        ["Membership", "N"],
        ascending=[True, False],
    )
)

display(concordance_class_df)
```

```python
# ============================================
# 26. SSF selected-species PR-effect advantage
# ============================================

rows = []

for _, r in ssf_df.iterrows():
    selected = r["Membership"]
    others = [s for s in SPECIES_ORDER if s != selected]

    selected_effect = abs(r[f"{selected}_Rho"])
    other_effects = [abs(r[f"{s}_Rho"]) for s in others]

    rows.append({
        "Feature": r["Feature"],
        "Membership": selected,
        "PR_class": r["PR_class"],
        "Selected_abs_rho": selected_effect,
        "Other_mean_abs_rho": np.mean(other_effects),
        "Effect_advantage": selected_effect - np.mean(other_effects),
    })

ssf_advantage_df = pd.DataFrame(rows)

# Summary
advantage_summary_df = (
    ssf_advantage_df
    .groupby("Membership")["Effect_advantage"]
    .agg(["count", "median", "mean"])
    .reset_index()
)

display(advantage_summary_df)

# Individual features
display(
    ssf_advantage_df
    .sort_values("Effect_advantage", ascending=False)
    .reset_index(drop=True)
)

# Plot
fig, ax = make_fig(w=4.5, h=4)

plot_data = [
    ssf_advantage_df.loc[
        ssf_advantage_df["Membership"].eq(sp),
        "Effect_advantage",
    ].to_numpy()
    for sp in SPECIES_ORDER
]

ax.boxplot(
    plot_data,
    tick_labels=SPECIES_ORDER,
    showfliers=False,
)

for i, sp in enumerate(SPECIES_ORDER, start=1):
    y = ssf_advantage_df.loc[
        ssf_advantage_df["Membership"].eq(sp),
        "Effect_advantage",
    ]

    x = np.full(len(y), i, dtype=float)
    x += np.random.default_rng(42 + i).normal(0, 0.04, len(y))

    ax.scatter(
        x,
        y,
        alpha=0.7,
        s=25,
    )

ax.axhline(
    0,
    color="black",
    linestyle="--",
    linewidth=1,
)

ax.set_xlabel("SSF membership")
ax.set_ylabel(
    r"Selected-species $|\rho|$ − mean other-species $|\rho|$"
)

plt.show()
```

```python
# ============================================
# 27. Test SSF PR-effect advantage
# ============================================

from scipy.stats import wilcoxon

rows = []

for group in ["All"] + SPECIES_ORDER:
    x = (
        ssf_advantage_df["Effect_advantage"]
        if group == "All"
        else ssf_advantage_df.loc[
            ssf_advantage_df["Membership"].eq(group), "Effect_advantage"
        ]
    )

    stat, p = wilcoxon(x, alternative="greater")

    rows.append({
        "Membership": group,
        "N": len(x),
        "Median_advantage": x.median(),
        "Mean_advantage": x.mean(),
        "N_positive": (x > 0).sum(),
        "Percent_positive": (x > 0).mean() * 100,
        "P_value": p,
    })

advantage_test_df = pd.DataFrame(rows)
advantage_test_df["FDR"] = multipletests(
    advantage_test_df["P_value"], method="fdr_bh"
)[1]

display(advantage_test_df)

# ============================================
# 28. Compare SSF PR-effect advantage
# ============================================

from scipy.stats import kruskal, mannwhitneyu

groups = {
    s: ssf_advantage_df.loc[
        ssf_advantage_df["Membership"].eq(s), "Effect_advantage"
    ]
    for s in SPECIES_ORDER
}

H, p = kruskal(*groups.values())

rows = []
for s1, s2 in combinations(SPECIES_ORDER, 2):
    stat, pv = mannwhitneyu(groups[s1], groups[s2], alternative="two-sided")
    rows.append({"Species_pair": f"{s1}-{s2}", "U": stat, "P_value": pv})

advantage_pairwise_df = pd.DataFrame(rows)
advantage_pairwise_df["FDR"] = multipletests(
    advantage_pairwise_df["P_value"], method="fdr_bh"
)[1]

print(f"Kruskal-Wallis: H = {H:.3f}, p = {p:.4g}")
display(advantage_pairwise_df)
```

```python
# ============================================
# 29. PR class vs SSF effect advantage
# ============================================

order = [
    "Conserved", "Quantitatively divergent", "Partially shared",
    "Opposite direction", "Species-restricted", "No clear pattern",
]

adv_class_df = (
    ssf_advantage_df.groupby("PR_class")["Effect_advantage"]
    .agg(N="size", Median="median", Mean="mean")
    .reindex(order).dropna().reset_index()
)

positive = (
    ssf_advantage_df.assign(Positive=lambda x: x["Effect_advantage"] > 0)
    .groupby("PR_class")["Positive"].sum()
)
adv_class_df["N_positive"] = adv_class_df["PR_class"].map(positive).astype(int)
adv_class_df["Percent_positive"] = 100 * adv_class_df["N_positive"] / adv_class_df["N"]

display(adv_class_df)

fig, ax = make_fig(w=7, h=3.8)
rng = np.random.default_rng(42)

data = [
    ssf_advantage_df.loc[ssf_advantage_df["PR_class"].eq(c), "Effect_advantage"].to_numpy()
    for c in adv_class_df["PR_class"]
]

ax.boxplot(data, tick_labels=adv_class_df["PR_class"], showfliers=False, widths=0.55)

for i, (c, y) in enumerate(zip(adv_class_df["PR_class"], data), start=1):
    x = i + rng.normal(0, 0.045, len(y))
    ax.scatter(x, y, s=24, alpha=0.7)
    ax.text(i, ax.get_ylim()[1], f"n={len(y)}", ha="center", va="bottom", fontsize=7)

ax.axhline(0, color="black", ls="--", lw=1)
ax.set_ylabel(r"Selected-species $|\rho|$ − mean other-species $|\rho|$")
ax.tick_params(axis="x", rotation=30)
plt.tight_layout()
plt.show()
```

```python
# ============================================
# 30. Compare SSF effect advantage by PR class
# ============================================

from scipy.stats import kruskal, mannwhitneyu

order = [
    "Conserved", "Quantitatively divergent", "Partially shared",
    "Opposite direction", "Species-restricted", "No clear pattern",
]

groups = {
    c: ssf_advantage_df.loc[
        ssf_advantage_df["PR_class"].eq(c), "Effect_advantage"
    ].dropna()
    for c in order
}

H, p = kruskal(*groups.values())
print(f"Kruskal-Wallis: H = {H:.3f}, p = {p:.4g}")

if p < 0.05:
    rows = []
    for c1, c2 in combinations(order, 2):
        u, pv = mannwhitneyu(groups[c1], groups[c2], alternative="two-sided")
        rows.append({"Class_1": c1, "Class_2": c2, "U": u, "P_value": pv})

    advantage_class_pairwise_df = pd.DataFrame(rows)
    advantage_class_pairwise_df["FDR"] = multipletests(
        advantage_class_pairwise_df["P_value"], method="fdr_bh"
    )[1]

    display(
        advantage_class_pairwise_df
        .sort_values("FDR")
        .reset_index(drop=True)
    )
```

```python
# ============================================
# 31. Composition of quantitatively divergent SSFs
# ============================================

# --------------------------------------------
# Add feature metadata and species PR effects
# back to the SSF advantage table
# --------------------------------------------

meta_cols = [
    "Feature",
    "Region",
    "Feature_type",
    "AT_Rho",
    "NB_Rho",
    "OS_Rho",
]

ssf_detail_df = (
    ssf_advantage_df
    .merge(
        pr_evidence_df[meta_cols].drop_duplicates("Feature"),
        on="Feature",
        how="left",
        validate="one_to_one",
    )
)

# Sanity checks
print("SSF rows:", len(ssf_detail_df))
print("Missing Region:", ssf_detail_df["Region"].isna().sum())
print("Missing Feature_type:", ssf_detail_df["Feature_type"].isna().sum())
print()


# --------------------------------------------
# Quantitatively divergent SSFs
# --------------------------------------------

qd_df = (
    ssf_detail_df.loc[
        ssf_detail_df["PR_class"].eq("Quantitatively divergent")
    ]
    .copy()
    .sort_values("Effect_advantage", ascending=False)
)

print(f"Quantitatively divergent SSFs: {len(qd_df)}")
print(
    f"Positive effect advantage: "
    f"{(qd_df['Effect_advantage'] > 0).sum()}/{len(qd_df)} "
    f"({100 * (qd_df['Effect_advantage'] > 0).mean():.1f}%)"
)
print()


# --------------------------------------------
# Individual features
# --------------------------------------------

qd_features_df = qd_df[
    [
        "Feature",
        "Region",
        "Feature_type",
        "Membership",
        "AT_Rho",
        "NB_Rho",
        "OS_Rho",
        "Selected_abs_rho",
        "Other_mean_abs_rho",
        "Effect_advantage",
    ]
].reset_index(drop=True)

display(qd_features_df)


# --------------------------------------------
# Feature-type composition
# QD vs all other SSFs
# --------------------------------------------

feature_type_counts = pd.crosstab(
    ssf_detail_df["Feature_type"],
    ssf_detail_df["PR_class"].eq("Quantitatively divergent"),
)

feature_type_counts = feature_type_counts.rename(
    columns={
        False: "Other_PR_classes",
        True: "Quantitatively_divergent",
    }
)

# Make robust in case one category is absent
for col in ["Other_PR_classes", "Quantitatively_divergent"]:
    if col not in feature_type_counts.columns:
        feature_type_counts[col] = 0

feature_type_counts["Total"] = (
    feature_type_counts["Other_PR_classes"]
    + feature_type_counts["Quantitatively_divergent"]
)

feature_type_counts["QD_percent"] = (
    100
    * feature_type_counts["Quantitatively_divergent"]
    / feature_type_counts["Total"]
)

feature_type_counts = (
    feature_type_counts[
        [
            "Quantitatively_divergent",
            "Other_PR_classes",
            "Total",
            "QD_percent",
        ]
    ]
    .sort_values(
        ["Quantitatively_divergent", "QD_percent"],
        ascending=False,
    )
)

display(feature_type_counts)


# --------------------------------------------
# Region composition
# QD vs all other SSFs
# --------------------------------------------

region_counts = pd.crosstab(
    ssf_detail_df["Region"],
    ssf_detail_df["PR_class"].eq("Quantitatively divergent"),
)

region_counts = region_counts.rename(
    columns={
        False: "Other_PR_classes",
        True: "Quantitatively_divergent",
    }
)

for col in ["Other_PR_classes", "Quantitatively_divergent"]:
    if col not in region_counts.columns:
        region_counts[col] = 0

region_counts["Total"] = (
    region_counts["Other_PR_classes"]
    + region_counts["Quantitatively_divergent"]
)

region_counts["QD_percent"] = (
    100
    * region_counts["Quantitatively_divergent"]
    / region_counts["Total"]
)

region_counts = (
    region_counts[
        [
            "Quantitatively_divergent",
            "Other_PR_classes",
            "Total",
            "QD_percent",
        ]
    ]
    .sort_values(
        ["Quantitatively_divergent", "QD_percent"],
        ascending=False,
    )
)

display(region_counts)
```

```python
# ============================================
# 32. PR-response class by transcript region
# ============================================

region_order = [
    "5'UTR",
    "CDS",
    "3'UTR",
    "mRNA",
]

pr_class_order = [
    "Conserved",
    "Quantitatively divergent",
    "Partially shared",
    "Opposite direction",
    "Species-restricted",
    "No clear pattern",
]


# --------------------------------------------
# Counts
# --------------------------------------------

region_pr_counts = pd.crosstab(
    ssf_detail_df["Region"],
    ssf_detail_df["PR_class"],
)

region_pr_counts = region_pr_counts.reindex(
    index=region_order,
    columns=pr_class_order,
    fill_value=0,
)

display(region_pr_counts)


# --------------------------------------------
# Within-region percentages
# --------------------------------------------

region_pr_pct = (
    region_pr_counts
    .div(region_pr_counts.sum(axis=1), axis=0)
    * 100
)

display(region_pr_pct.round(1))


# --------------------------------------------
# Compact summary
# --------------------------------------------

region_summary_df = (
    ssf_detail_df
    .groupby(["Region", "PR_class"])
    .size()
    .rename("N_features")
    .reset_index()
)

region_total = (
    ssf_detail_df
    .groupby("Region")
    .size()
    .rename("Region_total")
)

region_summary_df["Region_total"] = (
    region_summary_df["Region"]
    .map(region_total)
)

region_summary_df["Percent"] = (
    100
    * region_summary_df["N_features"]
    / region_summary_df["Region_total"]
)

region_summary_df["Region"] = pd.Categorical(
    region_summary_df["Region"],
    categories=region_order,
    ordered=True,
)

region_summary_df["PR_class"] = pd.Categorical(
    region_summary_df["PR_class"],
    categories=pr_class_order,
    ordered=True,
)

region_summary_df = (
    region_summary_df
    .sort_values(["Region", "PR_class"])
    .reset_index(drop=True)
)

display(region_summary_df)


# --------------------------------------------
# Plot
# --------------------------------------------

fig, axes = plt.subplots(
    1,
    2,
    figsize=(8, 3.5),
    dpi=_PLOT_CFG["dpi"],
)


# Counts
region_pr_counts.plot(
    kind="bar",
    stacked=True,
    ax=axes[0],
)

axes[0].set(
    title="Counts",
    xlabel="Transcript region",
    ylabel="Number of SSFs",
)

axes[0].tick_params(
    axis="x",
    rotation=0,
)

axes[0].get_legend().remove()


# Percentages
region_pr_pct.plot(
    kind="bar",
    stacked=True,
    ax=axes[1],
)

axes[1].set(
    title="Within-region composition",
    xlabel="Transcript region",
    ylabel="Percent",
)

axes[1].set_ylim(0, 100)

axes[1].tick_params(
    axis="x",
    rotation=0,
)

axes[1].legend(
    title="PR response",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    frameon=False,
)


plt.tight_layout()
plt.show()
```

```python
# ============================================
# 33. Region × PR-response association
# ============================================

from scipy.stats import chi2_contingency

# --------------------------------------------
# Observed contingency table
# --------------------------------------------

region_pr_test = pd.crosstab(
    ssf_detail_df["Region"],
    ssf_detail_df["PR_class"],
).reindex(
    index=region_order,
    columns=pr_class_order,
    fill_value=0,
)

display(region_pr_test)


# --------------------------------------------
# Classical chi-square
# Mainly used here as the test statistic
# --------------------------------------------

chi2_obs, p_chi2, dof, expected = chi2_contingency(
    region_pr_test,
    correction=False,
)

expected_df = pd.DataFrame(
    expected,
    index=region_pr_test.index,
    columns=region_pr_test.columns,
)

print(f"Chi-square = {chi2_obs:.3f}")
print(f"df = {dof}")
print(f"Asymptotic p = {p_chi2:.4g}")
print(
    "Expected cells < 5:",
    (expected_df < 5).sum().sum(),
    "/",
    expected_df.size,
)

display(expected_df.round(2))


# --------------------------------------------
# Permutation test
#
# Keep Region fixed and randomly permute
# PR-class labels among the 53 SSFs.
# --------------------------------------------

rng = np.random.default_rng(42)

n_perm = 10000

regions = ssf_detail_df["Region"].to_numpy()
classes = ssf_detail_df["PR_class"].to_numpy()

perm_stats = np.empty(n_perm)

for i in range(n_perm):

    permuted_classes = rng.permutation(classes)

    perm_table = pd.crosstab(
        pd.Categorical(
            regions,
            categories=region_order,
        ),
        pd.Categorical(
            permuted_classes,
            categories=pr_class_order,
        ),
        dropna=False,
    )

    perm_stats[i] = chi2_contingency(
        perm_table,
        correction=False,
    )[0]


# Add-one correction for permutation p-value
p_perm = (
    1 + np.sum(perm_stats >= chi2_obs)
) / (
    n_perm + 1
)

print()
print(f"Permutation test ({n_perm:,} permutations)")
print(f"Observed chi-square = {chi2_obs:.3f}")
print(f"Permutation p = {p_perm:.4g}")


# --------------------------------------------
# Standardized residuals
#
# Positive = more SSFs than expected
# Negative = fewer SSFs than expected
# --------------------------------------------

residuals = (
    region_pr_test - expected_df
) / np.sqrt(expected_df)

display(residuals.round(2))


# --------------------------------------------
# Largest deviations from expectation
# --------------------------------------------

residual_long_df = (
    residuals
    .stack()
    .rename("Standardized_residual")
    .reset_index()
)

residual_long_df["Abs_residual"] = (
    residual_long_df["Standardized_residual"].abs()
)

residual_long_df = (
    residual_long_df
    .sort_values(
        "Abs_residual",
        ascending=False,
    )
    .reset_index(drop=True)
)

display(residual_long_df)
```

```python
# ============================================
# 34. SSF effect advantage by transcript region
# ============================================

from scipy.stats import kruskal, mannwhitneyu
from statsmodels.stats.multitest import multipletests
from itertools import combinations


region_order = [
    "5'UTR",
    "CDS",
    "3'UTR",
    "mRNA",
]


# --------------------------------------------
# Summary
# --------------------------------------------

region_advantage_df = (
    ssf_detail_df
    .groupby("Region")["Effect_advantage"]
    .agg(
        N="size",
        Median="median",
        Mean="mean",
    )
    .reindex(region_order)
    .reset_index()
)

positive_counts = (
    ssf_detail_df
    .assign(
        Positive=ssf_detail_df["Effect_advantage"] > 0
    )
    .groupby("Region")["Positive"]
    .sum()
    .reindex(region_order)
)

region_advantage_df["N_positive"] = (
    positive_counts.to_numpy()
)

region_advantage_df["Percent_positive"] = (
    100
    * region_advantage_df["N_positive"]
    / region_advantage_df["N"]
)

display(region_advantage_df)


# --------------------------------------------
# Global Kruskal-Wallis test
# --------------------------------------------

groups = [
    ssf_detail_df.loc[
        ssf_detail_df["Region"].eq(region),
        "Effect_advantage",
    ].dropna()
    for region in region_order
]

H, p_kw = kruskal(*groups)

print(
    f"Kruskal-Wallis: "
    f"H = {H:.3f}, p = {p_kw:.4g}"
)


# --------------------------------------------
# Pairwise Mann-Whitney tests
# --------------------------------------------

pairwise_results = []

for r1, r2 in combinations(region_order, 2):

    x = ssf_detail_df.loc[
        ssf_detail_df["Region"].eq(r1),
        "Effect_advantage",
    ].dropna()

    y = ssf_detail_df.loc[
        ssf_detail_df["Region"].eq(r2),
        "Effect_advantage",
    ].dropna()

    U, p = mannwhitneyu(
        x,
        y,
        alternative="two-sided",
    )

    pairwise_results.append(
        {
            "Region_1": r1,
            "Region_2": r2,
            "N_1": len(x),
            "N_2": len(y),
            "U": U,
            "P_value": p,
        }
    )

region_pairwise_df = pd.DataFrame(
    pairwise_results
)

region_pairwise_df["FDR"] = multipletests(
    region_pairwise_df["P_value"],
    method="fdr_bh",
)[1]

region_pairwise_df = (
    region_pairwise_df
    .sort_values("P_value")
    .reset_index(drop=True)
)

display(region_pairwise_df)


# --------------------------------------------
# Plot
# --------------------------------------------

fig, ax = make_fig(
    w=6,
    h=3.8,
)

data = [
    ssf_detail_df.loc[
        ssf_detail_df["Region"].eq(region),
        "Effect_advantage",
    ].dropna()
    for region in region_order
]

bp = ax.boxplot(
    data,
    tick_labels=region_order,
    showfliers=False,
    patch_artist=True,
)

# Transparent boxes
for box in bp["boxes"]:
    box.set_facecolor("none")


# Individual observations
rng = np.random.default_rng(42)

for i, (region, values) in enumerate(
    zip(region_order, data),
    start=1,
):

    jitter = rng.normal(
        loc=i,
        scale=0.035,
        size=len(values),
    )

    ax.scatter(
        jitter,
        values,
        alpha=0.65,
        s=35,
    )

    # sample size
    ax.text(
        i,
        ax.get_ylim()[1],
        f"n={len(values)}",
        ha="center",
        va="bottom",
        fontsize=9,
    )


ax.axhline(
    0,
    color="black",
    linestyle="--",
    linewidth=1,
)

ax.set(
    xlabel="Transcript region",
    ylabel=(
        r"Selected-species $|\rho|$"
        "\n"
        r"$-$ mean other-species $|\rho|$"
    ),
)

plt.tight_layout()
plt.show()
```

```python
# ============================================
# 35. QD effect advantage within regions
# ============================================

from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests


region_order = [
    "5'UTR",
    "CDS",
    "3'UTR",
    "mRNA",
]


# --------------------------------------------
# QD vs all other PR classes
# --------------------------------------------

region_qd_df = ssf_detail_df.copy()

region_qd_df["QD_status"] = np.where(
    region_qd_df["PR_class"].eq("Quantitatively divergent"),
    "QD",
    "Other",
)


# --------------------------------------------
# Descriptive summary
# --------------------------------------------

region_qd_summary_df = (
    region_qd_df
    .groupby(
        ["Region", "QD_status"],
        observed=True,
    )["Effect_advantage"]
    .agg(
        N="size",
        Median="median",
        Mean="mean",
    )
    .reset_index()
)

region_qd_summary_df["Region"] = pd.Categorical(
    region_qd_summary_df["Region"],
    categories=region_order,
    ordered=True,
)

region_qd_summary_df = (
    region_qd_summary_df
    .sort_values(["Region", "QD_status"])
    .reset_index(drop=True)
)

display(region_qd_summary_df)


# --------------------------------------------
# Within-region QD vs Other
# --------------------------------------------

results = []

for region in region_order:

    sub = region_qd_df.loc[
        region_qd_df["Region"].eq(region)
    ]

    qd = sub.loc[
        sub["QD_status"].eq("QD"),
        "Effect_advantage",
    ].dropna()

    other = sub.loc[
        sub["QD_status"].eq("Other"),
        "Effect_advantage",
    ].dropna()

    # Do not test if one group is absent
    if len(qd) == 0 or len(other) == 0:
        U = np.nan
        p = np.nan
    else:
        U, p = mannwhitneyu(
            qd,
            other,
            alternative="two-sided",
        )

    results.append(
        {
            "Region": region,
            "N_QD": len(qd),
            "N_Other": len(other),
            "QD_median": qd.median() if len(qd) else np.nan,
            "Other_median": other.median() if len(other) else np.nan,
            "Median_difference": (
                qd.median() - other.median()
                if len(qd) and len(other)
                else np.nan
            ),
            "U": U,
            "P_value": p,
        }
    )


region_qd_test_df = pd.DataFrame(results)


# --------------------------------------------
# FDR only across tests that actually exist
# --------------------------------------------

valid = region_qd_test_df["P_value"].notna()

region_qd_test_df["FDR"] = np.nan

if valid.sum() > 0:
    region_qd_test_df.loc[valid, "FDR"] = multipletests(
        region_qd_test_df.loc[valid, "P_value"],
        method="fdr_bh",
    )[1]


display(region_qd_test_df)
```

```python
# ============================================
# 40. Redundancy among QD nucleotide features
# ============================================

from itertools import combinations
from scipy.stats import spearmanr


# --------------------------------------------
# QD nucleotide features
# --------------------------------------------

qd_kmer_features = (
    kmer_df.loc[
        kmer_df["QD_status"].eq("QD"),
        "Feature",
    ]
    .tolist()
)

print(f"QD nucleotide/k-mer features: {len(qd_kmer_features)}")


# --------------------------------------------
# Spearman correlation matrix
# --------------------------------------------

qd_kmer_corr = (
    x_raw[qd_kmer_features]
    .corr(method="spearman")
)

display(
    qd_kmer_corr.round(3)
)


# --------------------------------------------
# Pairwise correlations
# --------------------------------------------

pair_rows = []

for f1, f2 in combinations(qd_kmer_features, 2):

    pair = x_raw[[f1, f2]].dropna()

    rho, p = spearmanr(
        pair[f1],
        pair[f2],
    )

    pair_rows.append(
        {
            "Feature_1": f1,
            "Feature_2": f2,
            "N": len(pair),
            "Rho": rho,
            "Abs_rho": abs(rho),
            "P_value": p,
        }
    )


qd_kmer_pair_corr_df = (
    pd.DataFrame(pair_rows)
    .sort_values(
        "Abs_rho",
        ascending=False,
    )
    .reset_index(drop=True)
)

display(qd_kmer_pair_corr_df)


# --------------------------------------------
# Simple redundancy thresholds
# --------------------------------------------

for threshold in [0.7, 0.8, 0.9]:

    n_pairs = (
        qd_kmer_pair_corr_df["Abs_rho"] >= threshold
    ).sum()

    print(
        f"|rho| >= {threshold:.1f}: "
        f"{n_pairs} / {len(qd_kmer_pair_corr_df)} pairs"
    )
```

```python
# ============================================
# 41. Correlation structure among
#     QD nucleotide/k-mer SSFs
# ============================================

from scipy.cluster.hierarchy import linkage, leaves_list
from scipy.spatial.distance import squareform


# --------------------------------------------
# Distance from absolute Spearman correlation
# --------------------------------------------

abs_corr = qd_kmer_corr.abs()

distance = 1 - abs_corr

np.fill_diagonal(
    distance.values,
    0,
)


# --------------------------------------------
# Hierarchical clustering
# --------------------------------------------

linkage_matrix = linkage(
    squareform(
        distance.values,
        checks=False,
    ),
    method="average",
)

cluster_order = leaves_list(
    linkage_matrix
)

ordered_features = (
    abs_corr.index[cluster_order]
    .tolist()
)

print("Clustered feature order:")
for feature in ordered_features:
    print(feature)


# --------------------------------------------
# Reorder original signed correlations
# --------------------------------------------

corr_plot = qd_kmer_corr.loc[
    ordered_features,
    ordered_features,
]


# --------------------------------------------
# Plot
# --------------------------------------------

fig, ax = make_fig(
    w=6.5,
    h=5.5,
)

im = ax.imshow(
    corr_plot,
    vmin=-1,
    vmax=1,
    cmap="RdBu_r",
)

ax.set_xticks(
    range(len(ordered_features))
)

ax.set_yticks(
    range(len(ordered_features))
)

ax.set_xticklabels(
    ordered_features,
    rotation=45,
    ha="right",
)

ax.set_yticklabels(
    ordered_features,
)


# correlation values inside cells
for i in range(len(ordered_features)):
    for j in range(len(ordered_features)):

        value = corr_plot.iloc[i, j]

        ax.text(
            j,
            i,
            f"{value:.2f}",
            ha="center",
            va="center",
            fontsize=6,
        )


cbar = fig.colorbar(
    im,
    ax=ax,
)

cbar.set_label(
    "Spearman correlation"
)

ax.set_title(
    "Correlation among QD nucleotide/k-mer SSFs"
)

plt.tight_layout()
plt.show()
```

```python
# ============================================
# 42. Cross-species PR effects of
#     quantitatively divergent SSFs
# ============================================

qd_effect_df = (
    qd_df[
        [
            "Feature",
            "Membership",
            "Region",
            "Feature_type",
            "AT_Rho",
            "NB_Rho",
            "OS_Rho",
            "Effect_advantage",
        ]
    ]
    .copy()
    .sort_values("Effect_advantage", ascending=False)
    .reset_index(drop=True)
)

# Convert rho columns to numeric
rho_cols = ["AT_Rho", "NB_Rho", "OS_Rho"]

qd_effect_df[rho_cols] = (
    qd_effect_df[rho_cols]
    .apply(pd.to_numeric, errors="coerce")
)

display(qd_effect_df)


# --------------------------------------------
# Heatmap matrix
# --------------------------------------------

rho_matrix = (
    qd_effect_df
    .set_index("Feature")[rho_cols]
    .rename(
        columns={
            "AT_Rho": "AT",
            "NB_Rho": "NB",
            "OS_Rho": "OS",
        }
    )
)

rho_values = rho_matrix.to_numpy(dtype=float)


# --------------------------------------------
# Plot
# --------------------------------------------

fig, ax = make_fig(
    w=4.5,
    h=6,
)

vmax = np.nanmax(np.abs(rho_values))

im = ax.imshow(
    rho_values,
    cmap="RdBu_r",
    vmin=-vmax,
    vmax=vmax,
    aspect="auto",
)

ax.set_xticks(range(rho_matrix.shape[1]))
ax.set_xticklabels(rho_matrix.columns)

ax.set_yticks(range(rho_matrix.shape[0]))
ax.set_yticklabels(rho_matrix.index)


# Correlation values
for i in range(rho_matrix.shape[0]):
    for j in range(rho_matrix.shape[1]):
        value = rho_values[i, j]

        if np.isfinite(value):
            ax.text(
                j,
                i,
                f"{value:.2f}",
                ha="center",
                va="center",
                fontsize=7,
            )


cbar = fig.colorbar(im, ax=ax)
cbar.set_label(r"PR correlation ($\rho$)")

ax.set_xlabel("Species")
ax.set_ylabel("")
ax.set_title("Quantitatively divergent SSFs")

plt.tight_layout()
plt.show()
```
