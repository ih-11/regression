```python
# ============================================
# Global plotting configuration
# ============================================

_PLOT_CFG = {
    "fig_w": 6.0,
    "fig_h": 6.0,
    "dpi": 300,
}


SPECIES_INFO = {
    "AT21": {
        "label": "AT",
        "short": "AT21",
        "color": "#664D0AFF",
        "marker": "o",
    },
    "NB21": {
        "label": "NB",
        "short": "NB21",
        "color": "#7e3131",
        "marker": "^",
    },
    "OS21": {
        "label": "OS",
        "short": "OS21",
        "color": "#13563f",
        "marker": "s",
    },
}

def set_plot_style(
    *,
    base_fontsize=10,
    title_fontsize=12,
    label_fontsize=10,
    tick_fontsize=9,
    legend_fontsize=10,
    dpi=300,
    axes_linewidth=1.2,
    spines_top=True,
    spines_right=True,
    tick_size_major=6,
    tick_dir="out",
    grid=False,
    fig_w=6.0,
    fig_h=6.0,
):
    sns.set_style("ticks")

    mpl.rcParams.update({
        "font.family": "DejaVu Sans",
        "font.size": base_fontsize,

        "axes.titlesize": title_fontsize,
        "axes.labelsize": label_fontsize,

        "xtick.labelsize": tick_fontsize,
        "ytick.labelsize": tick_fontsize,

        "legend.fontsize": legend_fontsize,

        "figure.dpi": dpi,
        "savefig.dpi": dpi,

        "axes.linewidth": axes_linewidth,
        "axes.spines.top": spines_top,
        "axes.spines.right": spines_right,
        "axes.grid": grid,
        "axes.axisbelow": True,

        "xtick.major.size": tick_size_major,
        "ytick.major.size": tick_size_major,
        "xtick.direction": tick_dir,
        "ytick.direction": tick_dir,

        "legend.frameon": False,

        "savefig.bbox": "tight",
        "savefig.transparent": False,
        "figure.autolayout": False,
    })

    _PLOT_CFG.update({
        "fig_w": fig_w,
        "fig_h": fig_h,
        "dpi": dpi,
    })


def make_fig(w=None, h=None, dpi=None):
    W = float(w) if w is not None else _PLOT_CFG["fig_w"]
    H = float(h) if h is not None else _PLOT_CFG["fig_h"]
    D = dpi if dpi is not None else _PLOT_CFG["dpi"]

    fig, ax = plt.subplots(
        figsize=(W, H),
        dpi=D,
    )

    return fig, ax


def _compact_formatter():
    def _fmt(x, _pos=None):
        axx = abs(x)

        if axx >= 1e9:
            s = f"{x / 1e9:.1f}B"
        elif axx >= 1e6:
            s = f"{x / 1e6:.1f}M"
        elif axx >= 1e3:
            s = f"{x / 1e3:.1f}k"
        else:
            s = f"{x:.2g}"

        return (
            s.replace(".0B", "B")
             .replace(".0M", "M")
             .replace(".0k", "k")
        )

    return FuncFormatter(_fmt)


def format_axis(
    ax,
    *,
    xlabel=None,
    ylabel=None,
    compact_ticks=(),
):
    if xlabel is not None:
        ax.set_xlabel(xlabel)

    if ylabel is not None:
        ax.set_ylabel(ylabel)

    fmt = _compact_formatter()

    if "x" in compact_ticks:
        ax.xaxis.set_major_formatter(fmt)

    if "y" in compact_ticks:
        ax.yaxis.set_major_formatter(fmt)

    return ax


# ============================================
# Joint scatter with KDE marginals
# ============================================

def safe_pearsonr(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if len(x) < 2:
        return np.nan, np.nan

    if np.std(x) == 0 or np.std(y) == 0:
        return np.nan, np.nan

    return stats.pearsonr(x, y)


def safe_spearmanr(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if len(x) < 2:
        return np.nan, np.nan

    if np.std(x) == 0 or np.std(y) == 0:
        return np.nan, np.nan

    return stats.spearmanr(x, y)

def _kde_1d(values, lo, hi, num=256):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    grid = np.linspace(lo, hi, num)

    if len(values) < 2:
        return grid, np.zeros_like(grid)

    try:
        kde = stats.gaussian_kde(values)
        dens = kde(grid)
        dens /= dens.max() if dens.max() > 0 else 1
        return grid, dens

    except Exception:
        return grid, np.zeros_like(grid)


def joint_scatter(
    x,
    y,
    *,
    color=None,
    point_size=18,
    alpha=0.65,
    show_identity=True,
    show_regression=True,
    annotate=True,
    annotate_spearman=True,
    xlabel=None,
    ylabel=None,
    title=None,
    figsize=None,
    w=None,
    h=None,
    dpi=None,
    annotate_fontsize=14,
):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    n = len(x)

    if figsize is not None:
        FW, FH = figsize
    else:
        FW = float(w) if w is not None else _PLOT_CFG["fig_w"]
        FH = float(h) if h is not None else _PLOT_CFG["fig_h"]

    fig = plt.figure(
        figsize=(FW, FH),
        dpi=(dpi or _PLOT_CFG["dpi"]),
    )

    gs = GridSpec(
        2,
        2,
        width_ratios=(4, 1),
        height_ratios=(1, 4),
        hspace=0.05,
        wspace=0.05,
    )

    ax_top = fig.add_subplot(gs[0, 0])
    ax_joint = fig.add_subplot(gs[1, 0], sharex=ax_top)
    ax_right = fig.add_subplot(gs[1, 1], sharey=ax_joint)

    ax_joint.scatter(
        x,
        y,
        s=point_size,
        alpha=alpha,
        edgecolor="none",
        color=color,
    )

    lo = float(np.nanmin([x.min(), y.min()]))
    hi = float(np.nanmax([x.max(), y.max()]))

    pad = 0.05 * (hi - lo if hi > lo else 1.0)

    lo -= pad
    hi += pad

    ax_joint.set_xlim(lo, hi)
    ax_joint.set_ylim(lo, hi)

    if show_identity:
        ax_joint.plot(
            [lo, hi],
            [lo, hi],
            ls="--",
            lw=1.2,
            color="0.65",
            zorder=1,
        )

    if show_regression and n >= 2 and np.std(x) > 0 and np.std(y) > 0:
        slope, intercept = np.polyfit(x, y, 1)

        ax_joint.plot(
            [lo, hi],
            slope * np.array([lo, hi]) + intercept,
            color="black",
            lw=1.5,
            zorder=2,
        )

    format_axis(
        ax_joint,
        xlabel=xlabel,
        ylabel=ylabel,
        compact_ticks=(),
    )

    if title:
        ax_joint.set_title(title)

    if annotate and n >= 2 and np.std(x) > 0 and np.std(y) > 0:
        rp, _ = safe_pearsonr(x, y)
        rs, _ = safe_spearmanr(x, y)

        txt = rf"$r_p = {rp:.2f}$"

        if annotate_spearman:
            txt += "\n" + rf"$r_s = {rs:.2f}$"

        txt += f"\n$n = {n}$"

        ax_joint.text(
            0.04,
            0.96,
            txt,
            transform=ax_joint.transAxes,
            ha="left",
            va="top",
            fontsize=annotate_fontsize,
        )

    gx, dx = _kde_1d(x, lo, hi)
    gy, dy = _kde_1d(y, lo, hi)

    ax_top.plot(gx, dx, lw=2, color=color)
    ax_top.axis("off")

    ax_right.plot(dy, gy, lw=2, color=color)
    ax_right.axis("off")

    plt.tight_layout()

    return fig, (ax_joint, ax_top, ax_right)


set_plot_style()

# ============================================
# Plot helper functions
# ============================================

def species_palette():
    return [
        SPECIES_INFO["AT21"]["color"],
        SPECIES_INFO["NB21"]["color"],
        SPECIES_INFO["OS21"]["color"],
    ]


def outside_legend(
    ax,
    *,
    title="Species",
    n_items=3,
    x=1.02,
    y=1.00,
):
    handles, labels = ax.get_legend_handles_labels()

    ax.legend(
        handles[:n_items],
        labels[:n_items],
        title=title,
        frameon=False,
        loc="upper left",
        bbox_to_anchor=(x, y),
        borderaxespad=0,
    )

    return ax
```

```python
# ============================================
# 01. Load locked selected-feature dataset
# ============================================

RESULT_DIR = Path("/mnt/d/Ibnu/Programming/data/regression/results/lgbm")
SPECIES_KEYS = ["AT21", "NB21", "OS21"]
SPECIES_ORDER = ["AT", "NB", "OS"]

# Locked FI + SSS selection table
candidate_all_df = pd.read_csv(
    RESULT_DIR / "lgbm.species_specific_candidate_table.importance_p90.sss_p90.tsv.gz",
    sep="\t",
)

candidate_selected_df = (
    candidate_all_df
    .loc[candidate_all_df["Selected"]]
    .copy()
    .reset_index(drop=True)
)

candidate_features = sorted(candidate_selected_df["Feature"].unique())

feature_metadata_df = (
    candidate_selected_df[["Feature", "Region", "Feature_type"]]
    .drop_duplicates()
    .sort_values("Feature")
    .reset_index(drop=True)
)

# Raw biological data for the 61-feature union
raw_rows = []

for species_key in SPECIES_KEYS:
    species = SPECIES_INFO[species_key]["label"]

    with gzip.open(
        RESULT_DIR / f"{species_key}.lgbm.data_bundle.pkl.gz",
        "rb",
    ) as f:
        bundle = pickle.load(f)

    assert bundle["species"] == species_key
    assert set(candidate_features).issubset(bundle["feature_names"])

    metadata = pd.concat(
        [bundle["train_metadata"], bundle["test_metadata"]]
    )

    x_raw = pd.concat(
        [bundle["x_train_raw"], bundle["x_test_raw"]]
    )[candidate_features]

    assert metadata.index.equals(x_raw.index)

    temp = pd.concat([metadata, x_raw], axis=1)
    temp.insert(0, "Species", species)

    raw_rows.append(temp)

candidate_master_df = pd.concat(
    raw_rows,
    ignore_index=True,
)

# Sanity checks
selected_count_df = (
    candidate_selected_df
    .groupby("Species")
    .size()
    .reindex(SPECIES_ORDER)
    .rename("N_selected")
    .reset_index()
)

assert selected_count_df.set_index("Species")["N_selected"].to_dict() == {
    "AT": 19,
    "NB": 28,
    "OS": 23,
}

assert len(candidate_features) == 61
assert feature_metadata_df["Feature"].is_unique
assert set(feature_metadata_df["Feature"]) == set(candidate_features)
assert candidate_master_df[candidate_features].isna().sum().sum() == 0

print("Locked selected-feature dataset")
print("--------------------------------")
print("Selected rows   :", len(candidate_selected_df))
print("Unique features :", len(candidate_features))
print("Raw data shape  :", candidate_master_df.shape)

display(selected_count_df)
display(feature_metadata_df.head())
```

```python
# ============================================
# 02. Predictive selection membership
# SSF / PSF / MSF
# ============================================

membership_df = (
    candidate_selected_df.assign(Selected=1)
    .pivot_table(index="Feature", columns="Species", values="Selected", aggfunc="max", fill_value=0)
    .reindex(columns=SPECIES_ORDER, fill_value=0).reset_index()
    .merge(feature_metadata_df, on="Feature", how="left", validate="one_to_one")
)

membership_df["N_species"] = membership_df[SPECIES_ORDER].sum(axis=1).astype(int)
membership_df["Selection_class"] = membership_df["N_species"].map({1: "SSF", 2: "PSF", 3: "MSF"})
membership_df["Membership"] = membership_df.apply(
    lambda r: "-".join(sp for sp in SPECIES_ORDER if r[sp] == 1), axis=1
)

membership_df = (
    membership_df[
        ["Feature", "Region", "Feature_type", "AT", "NB", "OS",
         "N_species", "Selection_class", "Membership"]
    ]
    .sort_values(["N_species", "Membership", "Feature"], ascending=[False, True, True])
    .reset_index(drop=True)
)

class_summary_df = (
    membership_df["Selection_class"].value_counts()
    .reindex(["SSF", "PSF", "MSF"], fill_value=0)
    .rename("N_features").rename_axis("Selection_class").reset_index()
)

membership_summary_df = (
    membership_df.groupby(["Selection_class", "Membership"], as_index=False)
    .agg(N_features=("Feature", "size"))
)

assert len(membership_df) == 61
assert membership_df["Feature"].is_unique
assert membership_df["N_species"].between(1, 3).all()

print("Predictive selection membership")
display(class_summary_df)
display(membership_summary_df)
display(membership_df)
```

```python
# ============================================
# 03. Predictive vocabulary composition
# Region + feature type
# ============================================

region_order = ["5'UTR", "CDS", "3'UTR", "mRNA"]
type_order = ["Length", "RNA_structure_MFE", "Amino_acid_composition", "Nucleotide_kmer_freq"]
class_order = ["SSF", "PSF", "MSF"]

region_summary_df = (
    membership_df.groupby(["Selection_class", "Region"], as_index=False)
    .agg(N_features=("Feature", "size"))
)
region_summary_df["Percent"] = (
    100 * region_summary_df["N_features"]
    / region_summary_df.groupby("Selection_class")["N_features"].transform("sum")
)

type_summary_df = (
    membership_df.groupby(["Selection_class", "Feature_type"], as_index=False)
    .agg(N_features=("Feature", "size"))
)
type_summary_df["Percent"] = (
    100 * type_summary_df["N_features"]
    / type_summary_df.groupby("Selection_class")["N_features"].transform("sum")
)

overall_summary_df = pd.DataFrame({
    "N_features": [len(membership_df)],
    "N_SSF": [(membership_df["Selection_class"] == "SSF").sum()],
    "N_PSF": [(membership_df["Selection_class"] == "PSF").sum()],
    "N_MSF": [(membership_df["Selection_class"] == "MSF").sum()],
})

print("Predictive vocabulary composition")
display(overall_summary_df)
display(region_summary_df.sort_values(["Selection_class", "Region"]))
display(type_summary_df.sort_values(["Selection_class", "Feature_type"]))
```

```python
# ============================================
# 06. Feature–PR association across species
# Spearman rho + BH-FDR
# ============================================

from statsmodels.stats.multitest import multipletests

rows = []

for species in SPECIES_ORDER:
    d = candidate_master_df.query("Species == @species")

    for feature in candidate_features:
        rho, p = safe_spearmanr(d[feature], d["observed_raw"])
        rows.append({"Species": species, "Feature": feature, "PR_rho": rho, "P_value": p})

feature_pr_df = pd.DataFrame(rows)

feature_pr_df["P_adj"] = (
    feature_pr_df.groupby("Species")["P_value"]
    .transform(lambda x: multipletests(x, method="fdr_bh")[1])
)

feature_pr_df["Significant"] = feature_pr_df["P_adj"] < 0.05

feature_pr_df = (
    feature_pr_df
    .merge(
        membership_df[["Feature", "Region", "Feature_type", "Selection_class", "Membership"]],
        on="Feature", how="left", validate="many_to_one"
    )
    .sort_values(["Species", "PR_rho"], ascending=[True, False])
    .reset_index(drop=True)
)

pr_summary_df = (
    feature_pr_df.groupby("Species", as_index=False)
    .agg(
        N_features=("Feature", "size"),
        N_significant=("Significant", "sum"),
        Median_abs_rho=("PR_rho", lambda x: x.abs().median()),
        Max_abs_rho=("PR_rho", lambda x: x.abs().max()),
    )
)

print("Feature–PR association")
display(pr_summary_df)
display(feature_pr_df)
```

```python
# ============================================
# 07. Cross-species PR effect matrix
# 61 features × 3 species
# ============================================

pr_effect_df = (
    feature_pr_df
    .pivot(index="Feature", columns="Species", values="PR_rho")
    .reindex(columns=SPECIES_ORDER)
    .reset_index()
    .merge(
        membership_df[["Feature", "Region", "Feature_type", "Selection_class", "Membership"]],
        on="Feature", how="left", validate="one_to_one"
    )
)

pr_q_df = (
    feature_pr_df
    .pivot(index="Feature", columns="Species", values="P_adj")
    .reindex(columns=SPECIES_ORDER)
    .reset_index()
    .rename(columns={s: f"{s}_q" for s in SPECIES_ORDER})
)

pr_effect_df = (
    pr_effect_df
    .merge(pr_q_df, on="Feature", validate="one_to_one")
    .rename(columns={s: f"{s}_rho" for s in SPECIES_ORDER})
)

pr_effect_df["Max_abs_rho"] = pr_effect_df[
    [f"{s}_rho" for s in SPECIES_ORDER]
].abs().max(axis=1)

pr_effect_df = (
    pr_effect_df
    .sort_values("Max_abs_rho", ascending=False)
    .reset_index(drop=True)
)

assert len(pr_effect_df) == 61
assert pr_effect_df["Feature"].is_unique

display(
    pr_effect_df[
        ["Feature", "Region", "Feature_type", "Selection_class", "Membership",
         "AT_rho", "NB_rho", "OS_rho", "AT_q", "NB_q", "OS_q", "Max_abs_rho"]
    ]
)
```

```python
# ============================================
# 08. Global cross-species PR-effect conservation
# ============================================

from itertools import combinations

pairs = list(combinations(SPECIES_ORDER, 2))
point_color = "#8F6F6F"
rows = []

for s1, s2 in pairs:
    rho, p = safe_spearmanr(pr_effect_df[f"{s1}_rho"], pr_effect_df[f"{s2}_rho"])
    rows.append({"Species_pair": f"{s1}-{s2}", "N_features": len(pr_effect_df),
                 "Spearman_rho": rho, "P_value": p})

pr_effect_conservation_df = pd.DataFrame(rows)
display(pr_effect_conservation_df)

vals = pr_effect_df[[f"{s}_rho" for s in SPECIES_ORDER]].to_numpy(dtype=float)
lim = 1.08 * np.nanmax(np.abs(vals))

fig, axes = plt.subplots(
    1, 3, figsize=(7.5, 2.6), dpi=_PLOT_CFG["dpi"],
    sharex=True, sharey=True,
)

for ax, (s1, s2) in zip(axes, pairs):
    x, y = pr_effect_df[f"{s1}_rho"], pr_effect_df[f"{s2}_rho"]
    rho, p = safe_spearmanr(x, y)

    ax.scatter(x, y, s=20, color=point_color, alpha=0.75)
    ax.axhline(0, color="0.82", lw=0.8)
    ax.axvline(0, color="0.82", lw=0.8)

    p_text = f"{p:.2e}" if p < 0.001 else f"{p:.3f}"
    ax.text(
        0.05, 0.94,
        rf"Spearman $\rho$ = {rho:.2f}" + "\n" + rf"$p$ = {p_text}",
        transform=ax.transAxes, ha="left", va="top", fontsize=7.5,
    )

    ax.set_xlim(-lim, lim)
    ax.set_ylim(-lim, lim)
    ax.set_title(f"{s1} vs {s2}")
    ax.set_xlabel(f"{s1} PR effect")
    ax.set_ylabel(f"{s2} PR effect")

plt.tight_layout()
plt.show()
```

```python
# ============================================
# 09. Feature-level PR-effect divergence
# ============================================

rho_cols = [f"{s}_rho" for s in SPECIES_ORDER]

pr_divergence_df = pr_effect_df.copy()
pr_divergence_df["Effect_range"] = (
    pr_divergence_df[rho_cols].max(axis=1) - pr_divergence_df[rho_cols].min(axis=1)
)
pr_divergence_df["Effect_SD"] = pr_divergence_df[rho_cols].std(axis=1)
pr_divergence_df["Strongest_species"] = (
    pr_divergence_df[rho_cols].abs().idxmax(axis=1).str.replace("_rho", "", regex=False)
)
pr_divergence_df["Direction"] = pr_divergence_df[rho_cols].apply(
    lambda r: "Same" if len(set(np.sign(r))) == 1 else "Mixed", axis=1
)

pr_divergence_df = (
    pr_divergence_df
    .sort_values("Effect_range", ascending=False)
    .reset_index(drop=True)
)

display(
    pr_divergence_df[
        ["Feature", "Region", "Feature_type", "Selection_class", "Membership",
         "AT_rho", "NB_rho", "OS_rho", "Effect_range", "Effect_SD",
         "Strongest_species", "Direction"]
    ]
)
```
